# Se contrstruye el modelo

```python
import tensorflow as tf
from tensorflow.keras import layers, Model, Input
from sklearn.model_selection import train_test_split
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.losses import Huber

num_stations = len(le_station.classes_)
num_features = X_context.shape[1]

print(f"Number of stations: {num_stations}")
print(f"Number of features: {num_features}")

# ============================================================
# 4️⃣ Dividir datos en train/test
# ============================================================
(
    X_train, X_test,
    start_train, start_test,
    end_train, end_test,
    y_train, y_test
) = train_test_split(
    X_context, X_start, X_end, y,
    test_size=0.2,
    random_state=42
)

# ============================================================
# 5️⃣ Preparar los tensores de entrada
# ============================================================
X_train_context = X_train.values.astype(np.float32)
X_test_context  = X_test.values.astype(np.float32)

start_train_input = start_train.values.reshape(-1, 1)
start_test_input  = start_test.values.reshape(-1, 1)

end_train_input = end_train.values.reshape(-1, 1)
end_test_input  = end_test.values.reshape(-1, 1)

y_train = y_train.values.astype(np.float32)
y_test  = y_test.values.astype(np.float32)

# ============================================================
# 6️⃣ Construcción del modelo con embeddings
# ============================================================
embedding_dim = int(np.ceil(np.sqrt(num_stations)))

print(f"Embedding dimension: {embedding_dim}")

context_input = layers.Input(shape=(num_features,), name="context")
start_input   = layers.Input(shape=(1,), name="start_station_input")
end_input     = layers.Input(shape=(1,), name="end_station_input")

# Embeddings
start_embed = layers.Embedding(num_stations, embedding_dim, name='start_embedding')(start_input)
end_embed   = layers.Embedding(num_stations, embedding_dim, name='end_embedding')(end_input)

start_embed = layers.Flatten()(start_embed)
end_embed   = layers.Flatten()(end_embed)

# Concatenar todo
x = layers.Concatenate()([start_embed, end_embed, context_input])

x = layers.Dense(512, activation='relu')(x)
# x = layers.LeakyReLU()(x)
# x = layers.BatchNormalization()(x)
# x = layers.Dropout(0.1)(x)

x = layers.Dense(256, activation='relu')(x)
# x = layers.LeakyReLU()(x)
# x = layers.BatchNormalization()(x)
# x = layers.Dropout(0.1)(x)

x = layers.Dense(128, activation='relu')(x)
# x = layers.LeakyReLU()(x)
# x = layers.BatchNormalization()(x)
# x = layers.Dropout(0.1)(x)

x = layers.Dense(64, activation='relu')(x)
# x = layers.LeakyReLU()(x)
# x = layers.BatchNormalization()(x)
# x = layers.Dropout(0.1)(x)

output = layers.Dense(1, activation='linear')(x)
#output = layers.Dense(1, activation='softplus')(x)  # garantiza >0

model = Model(inputs=[start_input, end_input, context_input], outputs=output)
#model.compile(optimizer=Adam(1e-3), loss='mse', metrics=['mae']) # No funciona 
#model.compile(optimizer=Adam(1e-3), loss='mae', metrics=['mae']) #TODO
model.compile(optimizer=Adam(1e-3), loss=Huber(delta=1.0), metrics=['mae']) # Probadno, añadir un StandarScaler en lugar de sqrt
#model.compile(optimizer=Adam(1e-3), loss='poisson', metrics=['mae']) # TODO cambiar la salida a softplus
model.summary()
```

🔹 2️⃣ Por qué se usa en problemas de regresión como el tuyo

Tu variable objetivo (y) —probablemente número de viajes entre estaciones o similar—
tiene una distribución muy asimétrica (sesgada a la derecha):

muchos valores pequeños (pocos viajes),

unos pocos valores muy grandes (rutas muy transitadas).

Eso provoca que:

el modelo aprenda peor los valores pequeños,

y se concentre en minimizar los grandes errores de los valores enormes.

El logaritmo soluciona eso porque comprime los valores grandes y expande los pequeños,
haciendo la distribución más simétrica y estable numéricamente.

🔹 3️⃣ Ventajas directas
Problema sin log	Efecto del log1p
Distribución muy sesgada	La hace más normal (gaussiana)
Grandes diferencias entre valores	Reduce la escala y estabiliza
Gradientes inestables (explosivos)	Suaviza y acelera el entrenamiento
Pérdida dominada por valores grandes	Equilibra el error entre pequeños y grandes

```python
# y_train_scaled = np.log1p(y_train)
# y_test_scaled = np.log1p(y_test)

y_train_scaled = np.sqrt(y_train)
y_test_scaled = np.sqrt(y_test)
```

```python
from tensorflow.keras.callbacks import EarlyStopping

# ---------- Callbacks
callback = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)

# ---------- Entrenamiento
history = model.fit(
    [start_train_input, end_train_input, X_train_context],
    y_train_scaled,
    validation_data=([start_test_input, end_test_input, X_test_context], y_test_scaled),
    epochs=50,
    batch_size=64,
    callbacks=[callback],
    verbose=1
)
```

```python
test_loss, test_mae = model.evaluate(
    [start_test_input, end_test_input, X_test_context],
    y_test_scaled,
    verbose=0
)

print(f"Test Loss (MSE): {test_loss:.4f}")
print(f"Test MAE: {test_mae:.4f}")
```

# Métricas de Evaluación de Modelos

## 1️⃣ MAE (Mean Absolute Error)

Mide el error medio absoluto, es decir, cuánto se equivoca el modelo en promedio.

> Los valores “buenos” dependen de la escala de tu variable **y** (en tu caso, viajes por hora).

Interpretación aproximada si redondeas a múltiplos de 5:

| MAE   | Interpretación                                        |
|-------|------------------------------------------------------|
| 0–5   | Excelente: casi siempre acierta el múltiplo de 5    |
| 5–10  | Bueno: errores aceptables para predicciones agregadas|
| 10–20 | Regular: puede ser útil pero con cuidado            |
| >20   | Malo: predicciones poco confiables                  |

---

## 2️⃣ RMSE (Root Mean Squared Error)

Penaliza más los errores grandes que el MAE.

- Si **RMSE ≈ MAE** → errores uniformes y sin picos grandes.  
- Si **RMSE >> MAE** → hay predicciones muy equivocadas (outliers).

> Valores buenos dependen también de la escala de **y**.  
> Ideal: **RMSE ≤ 1×MAE** o ≤ 10 unidades (dependiendo del contexto).

---

## 3️⃣ R² (Coeficiente de determinación)

Mide qué proporción de la varianza de los datos reales explica el modelo.

| R²       | Interpretación                               |
|----------|---------------------------------------------|
| 0.8–1    | Muy bueno: predice casi todo                |
| 0.5–0.8  | Bueno: explica bastante variación           |
| 0–0.5    | Regular: predice algo pero no muy confiable|
| <0       | Malo: predice peor que la media            |

> 💡 En tus resultados anteriores, **R² ~ 0 o negativo** → el modelo no captura la tendencia real, aunque **MAE** sea pequeño porque predice algo cercano a la media.

---

## 4️⃣ Tiempo de inferencia

Medido en **milisegundos por muestra**.

| Tiempo     | Interpretación                                        |
|-----------|-------------------------------------------------------|
| <10 ms    | Excelente: ideal para producción en tiempo real      |
| 10–50 ms  | Bueno: aceptable para sistemas semi-reales           |
| 50–100 ms | Regular: puede ser suficiente según la frecuencia   |
| >100 ms   | Malo: demasiado lento para producción en tiempo real |

> En tus resultados, **63–107 ms por muestra** → regular, aceptable si no es crítico, pero puede mejorar si reduces embedding o batch-predict.
