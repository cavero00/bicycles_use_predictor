```python
# Agregar los viajes
df_counts = df_model.groupby(
    ['start_station_idx', 'end_station_idx', 'datetime_hour']
).agg({
    'temp_std': 'mean',
    'wind_std': 'mean',
    'rel_humidity_std': 'mean',
    'event': 'first',
    'day_type_Holiday': 'first',
    'day_type_Normal': 'first',
    'day_type_Weekend': 'first',
    'hour_float': 'first',
    'hour_sin': 'first',
    'hour_cos': 'first',
    'month_sin': 'first',
    'month_cos': 'first',
    'year': 'first',
    # cuenta de viajes
    'start_station_idx': 'count'
}).rename(columns={'start_station_idx': 'trip_count'}).reset_index()
```

```python
df_counts.iloc[10000:10010]
```

```python
X_context = df_counts[['temp_std', 'wind_std', 'rel_humidity_std', 'event', 'day_type_Holiday', 'day_type_Normal', 'day_type_Weekend', 'hour_float', 'hour_sin', 'hour_cos', 'month_sin', 'month_cos', 'year']]
X_start = df_counts['start_station_idx']
X_end = df_counts['end_station_idx']
y = df_counts['trip_count']
```

```python
with open("../data/normalized/le_station_encoder.pkl", "rb") as f:
    le_station = pickle.load(f)
    
len(le_station.classes_)
```

# Se contrstruye el modelo

```python
import tensorflow as tf
from tensorflow.keras import layers, Model, Input
from sklearn.model_selection import train_test_split
from tensorflow.keras.optimizers import Adam

num_stations = len(le_station.classes_)
num_features = X_context.shape[1]

print(f"Number of stations: {num_stations}")
print(f"Number of features: {num_features}")

# ============================================================
# 4️⃣ Dividir datos en train/test
# ============================================================
(
    X_train, X_test,
    start_train, start_test,
    end_train, end_test,
    y_train, y_test
) = train_test_split(
    X_context, X_start, X_end, y,
    test_size=0.2,
    random_state=42
)

# ============================================================
# 5️⃣ Preparar los tensores de entrada
# ============================================================
X_train_context = X_train.values.astype(np.float32)
X_test_context  = X_test.values.astype(np.float32)

start_train_input = start_train.values.reshape(-1, 1)
start_test_input  = start_test.values.reshape(-1, 1)

end_train_input = end_train.values.reshape(-1, 1)
end_test_input  = end_test.values.reshape(-1, 1)

y_train = y_train.values.astype(np.float32)
y_test  = y_test.values.astype(np.float32)

# ============================================================
# 6️⃣ Construcción del modelo con embeddings
# ============================================================
embedding_dim = int(np.ceil(np.sqrt(num_stations)))

print(f"Embedding dimension: {embedding_dim}")

context_input = layers.Input(shape=(num_features,), name="context")
start_input   = layers.Input(shape=(1,), name="start_station_input")
end_input     = layers.Input(shape=(1,), name="end_station_input")

# Embeddings
start_embed = layers.Embedding(num_stations, embedding_dim, name='start_embedding')(start_input)
end_embed   = layers.Embedding(num_stations, embedding_dim, name='end_embedding')(end_input)

start_embed = layers.Flatten()(start_embed)
end_embed   = layers.Flatten()(end_embed)

# Concatenar todo
x = layers.Concatenate()([start_embed, end_embed, context_input])
x = layers.Dense(256, activation='relu')(x)
x = layers.Dense(128, activation='relu')(x)
x = layers.Dense(64, activation='relu')(x)
output = layers.Dense(1, activation='linear')(x)

model = Model(inputs=[start_input, end_input, context_input], outputs=output)
model.compile(optimizer=Adam(1e-3), loss='mse', metrics=['mae'])
model.summary()
```

🔹 2️⃣ Por qué se usa en problemas de regresión como el tuyo

Tu variable objetivo (y) —probablemente número de viajes entre estaciones o similar—
tiene una distribución muy asimétrica (sesgada a la derecha):

muchos valores pequeños (pocos viajes),

unos pocos valores muy grandes (rutas muy transitadas).

Eso provoca que:

el modelo aprenda peor los valores pequeños,

y se concentre en minimizar los grandes errores de los valores enormes.

El logaritmo soluciona eso porque comprime los valores grandes y expande los pequeños,
haciendo la distribución más simétrica y estable numéricamente.

🔹 3️⃣ Ventajas directas
Problema sin log	Efecto del log1p
Distribución muy sesgada	La hace más normal (gaussiana)
Grandes diferencias entre valores	Reduce la escala y estabiliza
Gradientes inestables (explosivos)	Suaviza y acelera el entrenamiento
Pérdida dominada por valores grandes	Equilibra el error entre pequeños y grandes

```python
y_train_scaled = np.log1p(y_train)
y_test_scaled = np.log1p(y_test)
```

```python
from tensorflow.keras.callbacks import EarlyStopping

# ---------- Callbacks
callback = tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True)

# ---------- Entrenamiento
history = model.fit(
    [start_train_input, end_train_input, X_train_context],
    y_train_scaled,
    validation_data=([start_test_input, end_test_input, X_test_context], y_test_scaled),
    epochs=15,
    batch_size=128,
    callbacks=[callback],
    verbose=1
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import time
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# ------------------------------------------------------------
# 1️⃣ Predicciones en todo el conjunto de entrenamiento
# ------------------------------------------------------------
y_pred_train = model.predict(
    [start_train_input, end_train_input, X_train_context],
    verbose=0
)
y_pred_test = model.predict(
    [start_test_input, end_test_input, X_test_context],
    verbose=0
)

# Volver a la escala original (si entrenaste con log1p)
y_train_real = np.expm1(y_train_scaled)
y_train_pred_real = np.expm1(y_pred_train)
y_test_real = np.expm1(y_test_scaled)
y_test_pred_real = np.expm1(y_pred_test)

# ------------------------------------------------------------
# 2️⃣ Evaluación visual (comparar real vs predicho)
# ------------------------------------------------------------
plt.figure(figsize=(12, 5))
plt.plot(y_test_real[:200], label='Real', linewidth=2)
plt.plot(y_test_pred_real[:200], label='Predicho', linewidth=2)
plt.legend()
plt.title("Comparación predicciones vs valores reales (primeras 200 muestras de test)")
plt.xlabel("Muestras")
plt.ylabel("Viajes (trip_count)")
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()

# ------------------------------------------------------------
# 3️⃣ Métricas numéricas de desempeño
# ------------------------------------------------------------
mae_train = mean_absolute_error(y_train_real, y_train_pred_real)
rmse_train = np.sqrt(mean_squared_error(y_train_real, y_train_pred_real))
r2_train = r2_score(y_train_real, y_train_pred_real)

mae_test = mean_absolute_error(y_test_real, y_test_pred_real)
rmse_test = np.sqrt(mean_squared_error(y_test_real, y_test_pred_real))
r2_test = r2_score(y_test_real, y_test_pred_real)

print("=== Métricas Entrenamiento ===")
print(f"MAE  : {mae_train:.4f}")
print(f"RMSE : {rmse_train:.4f}")
print(f"R²   : {r2_train:.4f}")

print("\n=== Métricas Test ===")
print(f"MAE  : {mae_test:.4f}")
print(f"RMSE : {rmse_test:.4f}")
print(f"R²   : {r2_test:.4f}")

# ------------------------------------------------------------
# 4️⃣ Medición del tiempo de inferencia por muestra
# ------------------------------------------------------------
sample = [start_test_input[:1], end_test_input[:1], X_test_context[:1]]

t0 = time.perf_counter()
y_sample_pred = model.predict(sample, verbose=0)
t1 = time.perf_counter()

inference_time = (t1 - t0) * 1000  # milisegundos
print(f"\nTiempo de inferencia por muestra: {inference_time:.3f} ms")
```
