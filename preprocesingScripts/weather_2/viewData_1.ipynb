```python
import pandas as pd
import sys
import os
import pickle

sys.path.append(os.path.abspath("../../"))

from tools.tools import Tools

pd.set_option("display.max_rows", None)   # Muestra todas las filas
pd.set_option("display.max_columns", None)  # Muestra todas las columnas
pd.set_option("display.width", None)     # No corta la tabla en varias líneas
pd.set_option("display.max_colwidth", None)  # Muestra el contenido de celdas completo
```

```python
weather_path = r"C:\Users\burvu\Documents\Master IA\M10. TFM\data\weather"
years = ["2022", "2023", "2024"]

# Se leen los archvios CSV para cada año y se unifican en un solo DataFrame
data = dict()

for root, dirs, files in os.walk(weather_path):
    for dir_name in dirs: 
        data_per_year = dict()          
        print(f"-- {dir_name} --")
        if dir_name in years:
            data[dir_name] = dict()
            path_actual = os.path.join(root, dir_name)

            print(f"\tProcesando directorio: {path_actual}")
            
            # Busca y lee el psv
            for file in os.listdir(path_actual):
                if file.endswith(".psv"): 
                    print(f"\t\tLeyendo archivo: {file}")       
                    psv_path = os.path.join(path_actual, file)  
                    station_name = file.split("_")[1]
                    data_per_year[station_name] = pd.read_csv(psv_path, sep='|')
                    
            data[dir_name] = data_per_year
```

```python
cols = [
    "STATION",
    "Station_name",
    "DATE",
    "LATITUDE",
    "LONGITUDE",
    "temperature",
    "wind_speed",
    "precipitation",
    "relative_humidity",
    "snow_depth",
    "station_level_pressure"
]
```

```python
data_filtered = {}

for key, stations in data.items():
    data_filtered[key] = {}  # crea subdiccionario
    for station, df_station in stations.items():
        data_filtered[key][station] = df_station[cols].copy()
```

```python
for key, item in data_filtered.items():
    print(f"------- {key} -------")
    for station, df_station in item.items():
        print(f"\t--- {station} ---")
        print(df_station.describe())
        print(df_station.head())
        print(df_station.columns)
        print("\n")
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import math

for key, item in data_filtered.items():
    print(f"------- {key} -------")
    for station, df_station in item.items():
        print(f"\t--- {station} ---")
        
        # Convertir la columna DATE a tipo datetime
        df_station['DATE'] = pd.to_datetime(df_station['DATE'])
        
        # Configurar el tamaño de la figura
        plt.figure(figsize=(15, 10))
        
        # Número de columnas para los subplots
        num_cols = 2
        num_rows = math.ceil(len(cols[5:]) / num_cols)
        
        for i, col in enumerate(cols[5:]):
            plt.subplot(num_rows, num_cols, i + 1)
            plt.plot(df_station['DATE'], df_station[col], label=col)
            plt.xlabel('DATE')
            plt.ylabel(col)
            plt.title(f'{col} over Time')
            plt.legend()
            plt.grid()
        
        plt.tight_layout()
        plt.show()
```

```python
for year, stations in data_filtered.items():
    print(f"======= {year} =======")
    
    # Lista de estaciones y número
    station_names = list(stations.keys())
    num_stations = len(station_names)

    # Filtramos las columnas que queremos graficar (p. ej. a partir de la 6ª)
    vars_to_plot = cols[5:]
    num_vars = len(vars_to_plot)

    # Creamos figura: filas = variables, columnas = estaciones
    fig, axes = plt.subplots(num_vars, num_stations, figsize=(5*num_stations, 3*num_vars), sharex=False)
    
    # Si solo hay una estación o una variable, hacemos que axes sea 2D siempre
    if num_vars == 1 and num_stations == 1:
        axes = [[axes]]
    elif num_vars == 1:
        axes = [axes]
    elif num_stations == 1:
        axes = [[ax] for ax in axes]

    # Recorremos variables (filas)
    for row_idx, var in enumerate(vars_to_plot):
        # Recorremos estaciones (columnas)
        for col_idx, station in enumerate(station_names):
            df = stations[station].copy()
            df['DATE'] = pd.to_datetime(df['DATE'], errors='coerce')
            df = df.sort_values('DATE')

            ax = axes[row_idx][col_idx]
            ax.plot(df['DATE'], df[var], label=station, linewidth=1)
            ax.set_title(f"{station} - {var}")
            ax.set_xlabel('Fecha')
            ax.set_ylabel(var)
            ax.grid(True, linestyle='--', alpha=0.6)

            # Muestra la leyenda solo en la primera fila
            if row_idx == 0:
                ax.legend()

    plt.suptitle(f"Variables meteorológicas - {year}", fontsize=14, y=1.02)
    plt.tight_layout()
    plt.show()
```

```python
# Solo se escogen las estaciones de interés
stations_of_interest = ["USW00014819", "USW00094846", "USW00004807"]

data_filtered_interest = {}
for key, stations in data_filtered.items():
    data_filtered_interest[key] = {}  # crea subdiccionario
    for station, df_station in stations.items():
        print(station)
        if station in stations_of_interest:
            print(f"\tAdding station")
            data_filtered_interest[key][station] = df_station.copy()
```

```python
for key, item in data_filtered_interest.items():
    print(f"------- {key} -------")
    for station, df_station in item.items():
        print(f"\t--- {station} ---")
```

```python
import matplotlib.pyplot as plt
import pandas as pd

for year, stations in data_filtered_interest.items():
    print(f"======= {year} =======")

    station_names = list(stations.keys())
    num_stations = len(station_names)

    # Creamos una fila de subplots: una columna por estación
    fig, axes = plt.subplots(1, num_stations, figsize=(6 * num_stations, 4), sharex=False)
    
    # Si solo hay una estación, lo forzamos a lista para poder iterar
    if num_stations == 1:
        axes = [axes]

    for i, (station, df) in enumerate(stations.items()):
        df = df.copy()
        df['DATE'] = pd.to_datetime(df['DATE'], errors='coerce')
        df = df.sort_values('DATE').dropna(subset=['DATE'])

        # Calcula diferencia temporal en minutos (puedes cambiar a horas si prefieres)
        df['Δt_min'] = df['DATE'].diff().dt.total_seconds() / 60

        ax = axes[i]
        ax.plot(df['DATE'], df['Δt_min'], marker='o', markersize=3, linestyle='-')
        ax.set_title(f'{station}', fontsize=11)
        ax.set_xlabel('Fecha')
        ax.set_ylabel('Δt (min)')
        ax.grid(True, linestyle='--', alpha=0.6)

    plt.suptitle(f'Frecuencia de muestreo - {year}', fontsize=14, y=1.05)
    plt.tight_layout()
    plt.show()
```
